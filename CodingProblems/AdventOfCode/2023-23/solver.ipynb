```python
import time
```

```python
with open('data.txt', 'r') as file:
    lines = [[char for char in line.replace('\n', '')] for line in file]
width = len(lines[0])
height = len(lines)
start = (0, 1)
end = (width - 1, height - 2)
lines[start[0]][start[1]] = 'O'
```

```python
def print_grid(grid):
    for line in grid:
        for char in line:
            print(char, end = '')
        print('')
```

```python
slider = {'>': (0, 1),
          '<': (0, -1),
          '^': (-1, 0),
          'v': (1, 0)}
```

```python
start_time = time.time()
def get_available_directions(grid, row, column):
    if grid[row][column] in slider.keys():
        move = slider[grid[[row][column]]]
        if grid[row + move[0]][column + move[1]] in ('#', 'O'):
            return []
        else:
            return [move]
    directions = []
    if row != 0 and grid[row - 1][column] not in ('#', 'v', 'O'):
        directions.append((-1, 0))
    if row + 1 != width and grid[row + 1][column] not in ('#', '^', 'O'):
        directions.append((1, 0))
    if column != 0 and grid[row][column - 1] not in ('#', '>', 'O'):
        directions.append((0, -1))
    if column + 1 != height and grid[row][column + 1] not in ('#', '<', 'O'):
        directions.append((0, 1))
    return directions

def solve(grid, row, column, score):
    directions = get_available_directions(grid, row, column)
    if len(directions) == 0:
        return 0
    for direction in directions:
        if row + direction[0] == end[0] and column + direction[1] == end[1]:
            return score + 1
    max_score = 0
    for direction in directions:
        next_tile = grid[row + direction[0]][column + direction[1]]
        grid[row + direction[0]][column + direction[1]] = 'O'
        max_score = max(max_score, solve(grid, row + direction[0], column + direction[1], score + 1))
        grid[row + direction[0]][column + direction[1]] = next_tile
    return max_score

result = solve(lines, start[0], start[1], 0)
print((time.time() - start_time))
```
