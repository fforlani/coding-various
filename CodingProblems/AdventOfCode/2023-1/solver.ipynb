```python
import re
with open('data.txt', 'r') as file:
    lines = file.readlines()
```

```python
sum = 0
for line in lines:
    first = re.search("^[^\d]*(\d)", line)
    last = re.search("(\d)[^\d]*$", line)
    sum += int(first.group(1)) * 10 if first is not None else 0
    sum += int(last.group(1)) if last is not None else 0
sum
```
