```python
class Hint:
    
    def __init__(self, start, end, rowId, colId):
        self.start = start
        self.end = end
        self.rowId = rowId
        self.colId = colId
```

```python
maze = open('coding100/maze.txt', 'r')
text = maze.readlines()
maze.close()

hints = []
mapFile = open('coding100/map.txt', 'r')
lines = mapFile.read().splitlines()
counter = 0
while counter < len(lines):
    hints.append(Hint(lines[counter], lines[counter+4], int(lines[counter+1]), int(lines[counter+2])))
    counter += 6
mapFile.close()
```

```python
def findString(sub, startingIndex):
    for rowId in range(startingIndex, len(text)):
        try:
            if sub in text[rowId]:
                return [rowId, text[rowId].index(sub)]
        except:
            print("Error occurred with substrig: ", sub)
        
```

```python
answer = ""
for hint in hints:
    assert len(hint.start) == len(hint.end)
    
    rowIdStart, colIdStart = findString(hint.start, 0)
    rowIdEnd, colIdEnd = findString(hint.end, rowIdStart)
    if colIdStart != colIdEnd:
        print(hint.start, hint.end, rowIdStart, colIdStart, rowIdEnd, colIdEnd)
        continue
    assert colIdStart == colIdEnd
    subText = [row[colIdStart: colIdStart+len(hint.start)] for row in text[rowIdStart:rowIdEnd]]
    answer += subText[hint.rowId][hint.colId]
    
print(answer)
```
